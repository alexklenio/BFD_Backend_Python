**Exercício Prático:**

○ Crie uma classe Carro que represente um carro. O objetivo é abstrair as
funcionalidades essenciais para um motorista.

■ Requisitos:

● Atributos: marca, modelo, ano, e ligado (um booleano que
começa como False).

● Métodos:

- ligar(): deve mudar o atributo ligado para True e exibir
uma mensagem "O carro está ligado."
- desligar(): deve mudar o atributo ligado para False e
exibir "O carro está desligado."
- acelerar(valor): aumenta a velocidade em valor. O
carro precisa estar ligado para acelerar.
- frear(valor): diminui a velocidade em valor, sem
deixar a velocidade ficar negativa.
- exibir_informacoes(): deve exibir a marca, modelo e
ano do carro.

```python
class Carro():
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
    self.ligado = False
    self.velocidade = 0

    def ligar(self):
      self.ligado = True
      print("O carro está ligado")

    def desligar(self):
      self.ligado = False
      print("O carro está desligado")

    def acelerar(self, valor):
      if self.ligado:
        self.velocidade += valor
        print(f"Acelerando. Velocidade atual: {self.velocidade}")
      else:
        print("O carro precisa estar ligado para acelerar.")

    def frear(self, valor):
      if self.ligado:
        self.velocidade -= valor
        if self.velocidade < 0:
          self.velocidade = 0
        print(f"Freando. Velocidade atual: {self.velocidade}")

    def exibir_informacoes(self):
      print(f"Marca: {self.marca}, Modelo: {self.modelo}, Ano: {self.ano}")

    car = Carro("Toyota", "Corolla", 2022)
    car.ligar()
    car.acelerar(30)
    car.frear(10)
    car.desligar()
    car.exibir_informacoes()
```
