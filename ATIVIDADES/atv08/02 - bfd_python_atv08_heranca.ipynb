## ● **Exercício Teórico**

● Explique com suas palavras o que é herança e por que ela é útil na
programação. Dê um exemplo do mundo real (não relacionado a código)
de uma relação de herança (ex: "Animal" e "Cachorro").

Resposta:
<br>
<br>
Herânça é a capacidade de uma classe "herdar / ganhar" características e ações já existentes e definidas em outra classe, se tornando assim uma classe filha e outra clase exitente, sendo assim extremamente útil pois descarta a necessidade de repetição de codigo,uma vez que a classe "filha" possui por herança todos os atributos e métodos da classe "pai".
<br>
<br>
Exemplificando com a classe Animal, que por sua vez tem como atributos raça, sexo e cor do pelo, a classe Cachorro por herança possui os mesmos atributos mencionados anteriormente, mas sem a necessidade de implementar novamente um construtor.

<br>
● Verdadeiro ou Falso:
<br>

( F ) Uma classe filha herda apenas os métodos da classe pai, mas
não os atributos.

  ( V ) O principal benefício da herança é evitar a repetição de código.

  ( V ) Em Python, uma classe pode herdar de múltiplas classes pai.

## ● **Exercício Prático**

Crie um sistema simples para uma escola. Você terá uma classe pai
Pessoa e duas classes filhas Aluno e Professor.
<br>
<br>

■ Requisitos:
1. Classe Pessoa (Pai):
- Atributos: nome e idade.

- Método: exibir_dados(), que imprime o nome e a
idade.
<br>

2. Classe Aluno (Filha de Pessoa):

- Atributos: matricula.

- Método: exibir_dados(), que deve reutilizar o método
da classe pai e adicionar a matrícula.
<br>

3. Classe Professor (Filha de Pessoa):

- Atributos: disciplina.

- Método: exibir_dados(), que deve reutilizar o método
da classe pai e adicionar a disciplina.


```python
class Pessoa():
  def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade

    def exibir_dados(self):
      print(f'Nome: {self.nome} \nIdade: {self.idade}')

class Aluno(Pessoa):
  def __init__(self, nome, idade, matricula):
    super().__init__(nome, idade)
    self.matricula = matricula

  def exibir_dados(self):
    super().exibir_dados()
    print(f'Matricula: {self.matricula}')

class Professor(Pessoa):
  def __init__(self, nome, idade, disciplina):
    super().__init__(nome, idade)
    self.disciplina = disciplina

  def exibir_dados(self):
    super().exibir_dados()
    print(f'Disciplina: {self.disciplina}')
```
