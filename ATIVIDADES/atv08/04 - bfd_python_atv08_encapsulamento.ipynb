## ● **Exercício Prático**

○ Crie uma classe ContaBancaria.

■ Requisitos:

1. Atributo "privado": __saldo. O saldo inicial deve ser passado
no construtor.

2. Métodos Públicos:

- depositar(valor): Adiciona um valor ao saldo. Deve
validar se o valor é positivo.
- sacar(valor): Subtrai um valor do saldo. Deve validar
se o valor é positivo e se há saldo suficiente.
- get_saldo(): Retorna o valor do saldo atual. O acesso
direto a conta.__saldo deve ser evitado.

```python
class ContaBancaria:
  def __init__(self, saldo):
    self.__saldo = saldo

    def depositar(self, valor):
    if valor > 0:
      self.__saldo += valor

    def sacar(self, valor):
      if valor > 0 and valor <= self.__saldo:
        self.__saldo -= valor

    def get_saldo(self):
      return self.__saldo

conta = ContaBancaria(1000)
conta.depositar(500)
conta.sacar(200)
print(conta.get_saldo())
```
