## ● **Exercício Prático**

- Crie um sistema que pega uma lista de dados de usuários e os exporta
para diferentes formatos de arquivo (CSV e TXT). O sistema deve ser
flexível para que, no futuro, possamos adicionar novos formatos de
exportação sem alterar o código principal.

■ Requisitos:

● Crie uma classe base ExportadorDeDados. Ela deve ter:

- Um construtor que recebe o caminho_arquivo onde o
arquivo será salvo.
- Um método exportar(dados) que levanta um erro do
tipo NotImplementedError, forçando as classes filhas
a criarem suas próprias versões.

● Crie duas classes filhas que herdam de
ExportadorDeDados:
- ExportadorParaCSV: O método exportar deve
escrever os dados no formato CSV (valores
separados por vírgula).
- ExportadorParaTXT: O método exportar deve
escrever os dados em um formato de texto legível.

● Crie uma função gerar_relatorios(exportadores, dados) que
recebe uma lista de diferentes objetos exportadores e os
dados. Esta função deve chamar o método exportar de cada
um.

```python
class ExportadorDeDados:
    def __init__(self, caminho_arquivo):
        self.caminho_arquivo = caminho_arquivo

    def exportar(self, dados):
        raise NotImplementedError("Subclasse deve implementar este método.")

class ExportadorParaCSV(ExportadorDeDados):
    def exportar(self, dados):
        with open(self.caminho_arquivo, "w") as arquivo:
            for item in dados:
                linha = ",".join(map(str, item)) + "\n"
                arquivo.write(linha)
        print(f"Dados exportados para CSV em {self.caminho_arquivo}")

class ExportadorParaTXT(ExportadorDeDados):
    def exportar(self, dados):
        with open(self.caminho_arquivo, "w") as arquivo:
            for item in dados:
                arquivo.write(" | ".join(map(str, item)) + "\n")
        print(f"Dados exportados para TXT em {self.caminho_arquivo}")

def gerar_relatorios(exportadores, dados):
    for exportador in exportadores:
        exportador.exportar(dados)
```
